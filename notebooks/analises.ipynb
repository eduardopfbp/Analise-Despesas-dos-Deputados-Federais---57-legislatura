### Ajuste nas Colunas

```python
database_gastos = database_gastos.merge(
    database_deputados_id_legis_57[['id', 'nome', 'siglaPartido', 'siglaUf']],
    left_on='id_deputado',
    right_on='id',      
    how='left'  
)
```

```python
database_gastos['dataDocumento'] = pd.to_datetime(database_gastos['dataDocumento'])

database_gastos['dataDocumento_dd/mm/yyyy'] = database_gastos['dataDocumento'].dt.date
```

```python
database_gastos['ano_mes'] = (
    database_gastos['ano'].astype(str)
    + '-'
    + database_gastos['mes'].astype(str)
)
```

```python
database_gastos = database_gastos[['nome', 'id_deputado' , 'siglaPartido', 'siglaUf','tipoDespesa', 'tipoDocumento', 'valorDocumento',
       'codDocumento', 'codTipoDocumento', 'dataDocumento', 'numDocumento', 
       'urlDocumento', 'nomeFornecedor', 'cnpjCpfFornecedor', 'valorLiquido',
       'valorGlosa', 'numRessarcimento', 'codLote', 'parcela',
       'dataDocumento_dd/mm/yyyy', 'ano', 'mes', 'ano_mes']]
```

## Analise de Agrupamentos

### A grupamento por Deputado

```python
agrupado_por_deputado = (
    database_gastos.groupby(
        ['nome','id_deputado', 'siglaPartido', 'siglaUf'],
        as_index=False
    ).agg(
        total_gasto=('valorDocumento', 'sum'),
        media_gasto=('valorDocumento', 'mean'),
        qtd_notas=('numDocumento', 'count'),
        qtd_meses_com_gasto=('ano_mes', 'nunique')
    )
).round(2)

agrupado_por_deputado['media_mensal_gasto'] = (
    agrupado_por_deputado['total_gasto'] / agrupado_por_deputado['qtd_meses_com_gasto']
).round(2)


agrupado_por_deputado
```

```python
agrupado_por_deputado.sort_values('total_gasto', ascending=False).head(10)
```

### Agrupamento por Partido

```python
gastos_partido = (
    agrupado_por_deputado.groupby('siglaPartido')
    .agg({
        'total_gasto':'sum',
        'media_mensal_gasto':'mean',
        'id_deputado':'count'
    }).round(2)
    .rename(columns={'id_deputado':'qtd_deputados'})
).sort_values('total_gasto', ascending=False)

gastos_partido['media_gasto_por_deputado'] = (
    gastos_partido['total_gasto'] / gastos_partido['qtd_deputados']
).round(2)


gastos_partido.sort_values('media_gasto_por_deputado', ascending=False)
```

### Agrupamento por UF

```python
gastos_uf = (
    agrupado_por_deputado.groupby('siglaUf')
    .agg({
        'total_gasto':'sum',
        'media_mensal_gasto':'mean',
        'id_deputado':'count'
    }).round(2)
    .rename(columns={'id_deputado':'qtd_deputados'})
).sort_values('total_gasto', ascending=False)

gastos_uf['media_gasto_por_deputado'] = (
    gastos_uf['total_gasto'] / gastos_uf['qtd_deputados']
).round(2)


gastos_uf
```
